```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df= pd.read_csv("advertising_cost_vs_sales_500.csv")
df.head()
```

```python
x = df[["AdvertisingBudget"]]
y = df["Sales"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size = 0.2, random_state = 42
)
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_predict = model.predict(x_test)
print(y_predict)
```

```python
print("actual data: ",y_test)
print("predicted data: ",y_predict)
```

```python
new_value = pd.DataFrame({
    "AdvertisingBudget":[1000]
})

predict_new = model.predict(new_value)
print("new_result",predict_new[0])
```

```python
from sklearn.metrics import r2_score
r2_value = r2_score(y_test,y_predict)
print(r2_value) 
```

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test,y_predict)
print(mae)
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test,y_predict)
print(mse)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np 
mse = mean_squared_error(y_test,y_predict)
rmse = np.sqrt(mse)
print(rmse)
```

```python
import joblib
joblib.dump(model,"sales.pkl")
print("model saved successfully")
```
